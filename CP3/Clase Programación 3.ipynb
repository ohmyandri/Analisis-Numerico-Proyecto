```python
#Método de RK4 escalar

#Función
def f(x,y):
    return x+2*x*y
    
#Datos del problema
h = 0.5
x0 = 1
y0 = 3
xn = 3.5
n = int(np.round((xn-x0)/h,0))

for i in range(n):
    k1 = h*f(x0,y0)
    k2 = h*f(x0 + 0.5*h, y0 + 0.5*k1)
    k3 = h*f(x0 + 0.5*h, y0 + 0.5*k2)
    k4 = h*f(x0 + h, y0 + k3)
    y1 = y0 + (1/6)*(k1 + 2*k2 + 2*k3 + k4)
    y0 = y1
    x0 = x0+h

    print(f"Iteración {i+1}, con i={i}")
    print(f"k1 = {k1}")
    print(f"k2 = {k2}")
    print(f"k3 = {k3}")
    print(f"k4 = {k4}")
    print(f"y{i+1} = {y1} \n")
```
