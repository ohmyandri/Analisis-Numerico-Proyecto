```python
import numpy as np
import matplotlib.pyplot as plt

def dibujar_circulo(r, xc, yc):
    theta = np.linspace(0, 2*np.pi, 400)
    x = r * np.cos(theta) + xc
    y = r * np.sin(theta) + yc
    plt.plot(x, y)

plt.figure(figsize=(20,20))


num_circulos = 9830
for k in range(num_circulos):
    xc = (np.sin(np.pi*k/20000))**12*((1/2)*(np.cos(31*np.pi*k/10000))**16*np.sin(6*np.pi*k/10000)+(1/6)*(np.sin(31*np.pi*k/10000))**20) + 3*k/20000 + (np.cos(31*np.pi*k/10000))**6*np.sin((np.pi/2)*((k-10000)/10000)**7-np.pi/5)
    yc = (-9/4)*(np.cos(31*np.pi*k/10000))**6*np.cos(np.pi/2*((k-10000)/10000)**7-np.pi/5)*(2/3+(np.sin(np.pi*k/20000)*np.sin(3*np.pi*k/20000))**6) + (3/4)*(np.cos(3*np.pi*(k-10000)/100000))**10*(np.cos(9*np.pi*(k-10000)/100000))**10*(np.cos(36*np.pi*(k-10000)/100000))**14 + (7/10)*((k-10000)/10000)**2
    radio = (np.sin(np.pi*k/20000))**10*((1/4)*(np.cos(31*np.pi*k/10000+25*np.pi/32))**20+(1/20)*np.cos(31*np.pi*k/10000)**2) + (1/30)*(3/2-np.cos(62*np.pi*k/10000)**2)
    dibujar_circulo(radio, xc, yc)



plt.gca().set_aspect("equal", "box")
plt.show()

plt.savefig("Pajaro.png", dpi=300)
```
