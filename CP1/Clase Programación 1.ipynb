## Método de bisección

$$
x_m = \frac{a+b}{2}
$$

```python
a = 0
b = 2

max_iter = 500
epsilon = 1e-10
xa = a
cifras_decimales = 6

def f(x):
    return np.exp(x) + x - 2

for i in range(1, max_iter+1):
    print(f"Iteración {i}")
    xm = (a+b)/2
    if f(a)*f(xm) < 0:
        b = xm
        print("El cambio de signo está entre f(a) y f(xm)")
    elif f(xm)*f(b) < 0:
        a = xm
        print("El cambio de signo está entre f(xm) y f(b)")
    else:
        print(f"La raíz es xm = {xm}")
        break
    error = abs(xm - xa)
    xa = xm

    print(f"f(a) = {round(f(a),cifras_decimales)}, f(b) = {round(f(b),cifras_decimales)}, f(xm) = {round(f(xm),cifras_decimales)}")
    print(f"La raíz aproximada xm = {round(xm,cifras_decimales)}")
    print(f"El error absoluto aproximado ea = {round(error,cifras_decimales)} \n")
 
    if error < epsilon:
        break 
    
    
```

# Método de Newton - Raphson

```python
x0 = 1
max_iter = 50
epsilon = 1e-20
cifras_decimales = 6

def f(x):
    return np.exp(x) + x - 2

def fp(x):
    return np.exp(x) + 1
    
for i in range(1, max_iter+1):
    print(f"Iteración {i}")
    xi = x0 - f(x0)/fp(x0)
    error = abs(xi - x0)
    x0 = xi
    
    print(f"La raíz aproximada xi = {round(xi,cifras_decimales)}")
    print(f"El error absoluto aproximado ea = {round(error,cifras_decimales)} \n")
 
    if error < epsilon:
        break 
    
```
